```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('himeji.jpg')

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.title('oryginalny')
plt.imshow(img_gray, cmap='gray')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def apply_filter(grayscale_img, filter_kernel):
  image_height, image_width = grayscale_img.shape
  filtered_img = np.zeros((image_height - 2, image_width - 2), dtype=np.float32)
  for y in range(1, image_height - 1):
    for x in range(1, image_width - 1):
      ng = grayscale_img[y-1:y+2, x-1:x+2] #nasz mały filtr
      convolution_value = np.sum(ng * filter_kernel) #wartość konwolucji
      filtered_img[y-1,x-1] = convolution_value

  filtered_img = np.abs(filtered_img) #upewnienie się że wartości są całkowite (nie mają przecinka)
  filtered_img = np.clip(filtered_img, 0, 255).astype(np.uint8) #ucięcie wartości by mieściły się w zakresie 0-255

  plt.figure(figsize=(10,4))
  plt.subplot(1,2,1)
  plt.title('oryginalny')
  plt.imshow(img_gray, cmap='gray')

  plt.subplot(1,2,2)
  plt.title("Nałożony filtr")
  plt.imshow(filtered_img, cmap='gray')

  plt.show()
```

```python
apply_filter(img_gray, np.array([
    [-1,0,1],
    [-2,0,2],
    [-1,0,1]
]))
```
