```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# [10, 20, 35]

# zakres temperatur
x = np.linspace(0, 50, 500)

def fuzzy_warm(x):
  return np.maximum(np.minimum((x-10)/10, (35-x)/15), 0)


mu_warm = fuzzy_warm(x)

# charakterystyczne punkty (a,b,c)
points_x = [10, 20, 35]
points_y = fuzzy_warm(np.array(points_x))
labels = ["a(10)", "b(20)", "c(35)"]


plt.figure(figsize=(10, 5))
plt.plot(x, mu_warm, label="Ciepło", color="orange")

for i, label in enumerate(labels):
  plt.annotate(label, (points_x[i], points_y[i] + 0.05), ha='center')

plt.title("Funkcja przynależności: Ciepło (trójkatna: 10, 20, 35)")
plt.xlabel("Temperatura")
plt.ylabel("Stopień przynależności (0-1)")
plt.grid(True)
plt.legend()
plt.ylim(0, 1.1)
plt.show()
```
